```python
import torch
print(torch.mps.is_available())
```

```python
device = "mps" if torch.backends.mps.is_available() else "cpu"
print(f"Using device: {device}")   
```

```python
# we already have download Qwen3 
# if not downloaded 
from reasoning_from_scratch.qwen3 import download_qwen3_small
from reasoning_from_scratch.qwen3 import Qwen3Model, QWEN_CONFIG_06_B
# download_qwen3_small(kind="base", tokenizer_only=True, out_dir="./qwen3_model")
model = Qwen3Model(QWEN_CONFIG_06_B).to(device)
```

```python
from reasoning_from_scratch.qwen3 import Qwen3Tokenizer
from pathlib import Path
tokenizer_path = Path("./qwen3_model") / "tokenizer-base.json"
tokenizer = Qwen3Tokenizer(tokenizer_path)
```

```python
prompt = "explain quantum computing"
tokenized_prompt = tokenizer.encode(prompt)
print(tokenized_prompt)
print(tokenizer.decode(tokenized_prompt))
```

```python
model.to(device)
```

```python
# trying prompt through the model
prompt = "explain quantum computing"
tokenized_prompt = tokenizer.encode(prompt)
#print(torch.tensor(tokenized_prompt).unsqueeze(0))
input_tensor = torch.tensor(tokenized_prompt).unsqueeze(0).to(device)
with torch.inference_mode():
    output = model(input_tensor)

output = output.squeeze()
print(output.shape)
output_tokens = torch.argmax(output, dim=1)
print(tokenizer.decode(output_tokens.tolist()))
```
